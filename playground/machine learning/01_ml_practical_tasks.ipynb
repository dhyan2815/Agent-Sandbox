### Pandas Functions Reference Guide

A comprehensive reference for data manipulation, creation, deletion, and reshaping functions in pandas.

---

#### 📊 CREATION - Functions to create new DataFrames, Series, or data structures

| Function Name | Description | Example |
|---|---|---|
| `read_csv()` | Reads data from a CSV file into a DataFrame | `df = pd.read_csv("data.csv")` |
| `read_excel()` | Reads data from an Excel file into a DataFrame | `df = pd.read_excel("file.xlsx")` |
| `DataFrame()` | Creates a new DataFrame from lists, dicts, or arrays | `df = pd.DataFrame({"A": [1,2,3], "B": [4,5,6]})` |
| `Series()` | Creates a one-dimensional labeled array | `s = pd.Series([1,2,3], index=['a','b','c'])` |
| `copy()` | Creates a deep or shallow copy of a DataFrame | `df_copy = df.copy()` |

| Function Name | Description | Example |
|---|---|---|
| `insert()` | Inserts a new column at a specified position | `df.insert(1, "Age", 25)` |

| `set_index()` | Sets a column as the DataFrame's index | `df.set_index("id")` |
| `reset_index()` | Resets index back to default integer index | `df.reset_index()` |
| `astype()` | Converts a column's data type | `df["age"].astype(int)` |
| `apply()` | Applies a function along an axis of the DataFrame | `df.apply(np.sum, axis=0)` |
| `applymap()` | Applies a function element-wise to a DataFrame | `df.applymap(str)` |
| `map()` | Maps values of a Series using a function/dict | `df["status"].map({1: "one", 2: "two"})` |
| `transform()` | Applies a function returning same-shaped output | `df.transform(lambda x: x * 2)` |
| `sort_values()` | Sorts DataFrame by specified column values | `df.sort_values("age", ascending=False)` |
| `sort_index()` | Sorts DataFrame by index labels | `df.sort_index()` |

| `update()` | Updates DataFrame values using another DataFrame | `df.update(df2)` |
| `where()` | Replaces values where condition is False | `df.where(df > 0)` |
| `mask()` | Replaces values where condition is True | `df.mask(df < 0)` |

#### 💡 Usage Tips

##### When Creating DataFrames:
- Use `pd.read_csv()` for CSV files - fastest for large datasets
- Use `pd.DataFrame()` for creating from scratch with lists/dicts
- Always use `.copy()` when you need to preserve the original data

##### When Modifying Data:
- Use `fillna()` before analysis to handle missing values
- Use `astype()` to ensure correct data types for operations
- Chain operations using `.pipe()` or method chaining for readability
- `loc[]` is better for label-based access, `iloc[]` for position-based

##### When Reshaping:
- Use `groupby()` for aggregation and summary statistics
- Use `pivot()` for wide format, `melt()` for long format
- Use `merge()` for SQL-like joins between DataFrames

##### When Deleting:
- Use `drop()` for specific columns/rows
- Use `dropna()` to remove rows with any missing values
- Use `drop_duplicates()` after data cleaning

### Filling missing values - Given a DataFrame with missing values in a numeric column, fill the NaNs with the median of that column.

````python
def fill_miss_value(df):
  median_age = df['Age'].median() # calculate median
  df['Age'] = df['Age'].fillna(median_age) # fill NaN cells with the median values
  return df

filled = fill_miss_value(df)
print(filled)
````

### You have a dataframe. How do you handle missing values in the 'Age' column by replacing them with the mean

````python
def clean_dataset(df):
    mean_age = df['Age'].mean() # define the mean of the Age
    df['Age'] = df['Age'].fillna(mean_age) # filled empty cells with the mean value in the 'Age' column
    return df

cleaned = clean_dataset(df)
print(cleaned)
````

### Given a DataFrame of employees with columns Name, Department, and Salary, return the names of people in 'Engineering' earning more than 50,000, sorted by Salary descending.

````python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Department': ['Engineering', 'Sales', 'Engineering', 'Engineering', 'Marketing'],
    'Salary': [75000, 48000, 62000, 91000, 52000]
}

df = pd.DataFrame(data)

result = df[(df['Department'] == 'Engineering') & (df['Salary'] > 5000)].sort_values(by='Salary', ascending=False)
print(result)
````

### Group By Aggregation - Given a DataFrame with columns Category and Price, return the average price for each category

````python
data = {
    'Category' : ['A', 'B', 'A', 'D', 'B'],
    'Price' : [4500, 5600, 1300, 3500, 2500]
}

df = pd.DataFrame(data)

result = df.groupby('Category')['Price'].mean().reset_index()
result.head()
````

### Apply Function: Given a DataFrame with a Text column, create a new column Word_Count by applying a custom function that counts the words in Text

````python
import pandas as pd

# Sample DataFrame
data = {
    'Text': [
        'Hello world',
        'This is a sample sentence',
        'Another example with multiple words',
        'Short one'
    ]
}

df = pd.DataFrame(data)

df['Word_Count'] = df['Text'].str.split().str.len()
df

# first the text was split into whitespaces using - split() string in-built function
# then the count of word / length was calculated using - len() string in-built function
````

### One-Hot Encoding: Write a short script to convert a categorical column in a DataFrame into one-hot encoded columns using Pandas.

#### ⚠️**Hint** `get_dummies` function
- a pandas function that converts categorical variables into one-hot encoded binary columns

````python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'blue', 'red']
})

one_hot = pd.get_dummies(df['color'], prefix='color')
print(one_hot)
````

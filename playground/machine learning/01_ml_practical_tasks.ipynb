### Basic Pandas Execution (Very common for ML Freshers)
#### You have a dataframe. How do you handle missing values in the 'Age' column by replacing them with the mean

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({"Name": ["Alice", "Bob", "Charlie", "David", "Eva"], 
                    "Age": [25, np.nan, 30, np.nan, 45]})

def clean_dataset(df):
    mean_age = df['Age'].mean() # define the mean of the Age
    df['Age'] = df['Age'].fillna(mean_age) # filled empty cells with the mean value in the 'Age' column
    return df

cleaned = clean_dataset(df)
print(cleaned)
```

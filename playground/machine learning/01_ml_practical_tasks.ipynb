## Basic Pandas Execution (Very common for ML Freshers)


```python
import numpy as np
import pandas as pd

df = pd.DataFrame({"Name": ["Alice", "Bob", "Charlie", "David", "Eva"], "Age": [25, np.nan, 30, np.nan, 45]})
```

### Filling missing values - Given a DataFrame with missing values in a numeric column, fill the NaNs with the median of that column.

```python
def fill_miss_value(df):
  median_age = df['Age'].median() # calculate median
  df['Age'] = df['Age'].fillna(median_age) # fill NaN cells with the median values
  return df

filled = fill_miss_value(df)
print(filled)
```

### You have a dataframe. How do you handle missing values in the 'Age' column by replacing them with the mean

```python
def clean_dataset(df):
    mean_age = df['Age'].mean() # define the mean of the Age
    df['Age'] = df['Age'].fillna(mean_age) # filled empty cells with the mean value in the 'Age' column
    return df

cleaned = clean_dataset(df)
print(cleaned)
```

### Given a DataFrame of employees with columns Name, Department, and Salary, return the names of people in 'Engineering' earning more than 50,000, sorted by Salary descending.

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Department': ['Engineering', 'Sales', 'Engineering', 'Engineering', 'Marketing'],
    'Salary': [75000, 48000, 62000, 91000, 52000]
}

df = pd.DataFrame(data)

result = df[(df['Department'] == 'Engineering') & (df['Salary'] > 5000)].sort_values(by='Salary', ascending=False)
print(result)
```

### Group By Aggregation - Given a DataFrame with columns Category and Price, return the average price for each category

```python
data = {
    'Category' : ['A', 'B', 'A', 'D', 'B'],
    'Price' : [4500, 5600, 1300, 3500, 2500]
}

df = pd.DataFrame(data)

result = df.groupby('Category')['Price'].mean().reset_index()
result.head()
```

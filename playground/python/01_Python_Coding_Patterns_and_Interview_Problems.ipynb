### Python Coding Patterns & Interview Problems

#### Find all duplicate elements in a list.

```python
def find_duplicates(arr):
    seen = []
    duplicates = []
    for item in arr:
        if item in seen and item not in duplicates:
            duplicates.append(item)
        else:
            seen.append(item)
    return duplicates

list = [1, 2, 3, 4, 1, 2, 2, 5, 5, 4]
dupli_list = find_duplicates(list)

# Sort the list in ascending order
sorted(dupli_list)
```

#### Remove all duplicate elements in a list.
- Return a new list with all duplicates removed, preserving the original order

```python
og_list = [1, 2, 2, 3, 1, 4, 5, 5]
uni_list = []

for item in og_list:
    if item not in uni_list:
        uni_list.append(item)
print(uni_list)
```

#### Count the frequency of each Letter

- Using Library

```python
from collections import Counter

word = "The word taken is Banana"

count = Counter(word.lower())
print(count)
```

- Using without Library

```python
word = 'Banana'
letter_count = {}

for letter in word:
    if letter in letter_count:
        letter_count[letter] += 1
    else:
        letter_count[letter] = 1
print(letter_count)
```

#### Given a list of numbers, return a list of the squares.

```python
def squares(num):
    return [x**2 for x in num]
squares([1, 2, 3, 5, 6])
```

#### Given a list of numbers, return a list of the squares of only the even numbers.

```python
def sq_num(no):
    return [x**2 for x in no if x % 2 == 0]

sq_num([1, 2, 3, 4, 5, 6, 8, 10, 12, 14])
```

#### Operator based question

##### The rule for `and` is:

- If the first operand is truthy, return the second operand.
- If the first operand is falsy, return the first operand

##### The rule for `or` is:

- If the first operand is falsy, return the second operand.
- If the first operand is truthy, return the first operand.

```python
result = [1, 2] and "Hello"
print(result)
```

```python
result = 0 and 1 / 0
print(result)
```

```python
result = [] or "Hello"
print(result)
```

```python
result = 0 or (1, "hi", "dhyan")
print(result)
```

#### Find all duplicate elements in a list.

```python
def find_duplicates(arr):
    seen = []
    duplicates = []
    for item in arr:
        if item in seen and item not in duplicates:
            duplicates.append(item)
        else:
            seen.append(item)
    return duplicates

list = [1, 2, 3, 4, 1, 2, 2, 5, 5, 4]
dupli_list = find_duplicates(list)

# Sort the list in ascending order
sorted(dupli_list)
```

#### Remove all duplicate elements in a list.
- Return a new list with all duplicates removed, preserving the original order

```python
og_list = [1, 2, 2, 3, 1, 4, 5, 5]
uni_list = []

for item in og_list:
    if item not in uni_list:
        uni_list.append(item)
print(uni_list)
```

#### Count the frequency of each Letter

- Using Library

```python
from collections import Counter

word = "The word taken is Banana"

count = Counter(word.lower())
print(count)
```

- Using without Library

```python
word = 'Banana'
letter_count = {}

for letter in word:
    if letter in letter_count:
        letter_count[letter] += 1
    else:
        letter_count[letter] = 1
print(letter_count)
```

#### Given a list of numbers, return a list of the squares.

```python
def squares(num):
    return [x**2 for x in num]
squares([1, 2, 3, 5, 6])
```

#### Given a list of numbers, return a list of the squares of only the even numbers.

```python
def sq_num(no):
    return [x**2 for x in no if x % 2 == 0]

sq_num([1, 2, 3, 4, 5, 6, 8, 10, 12, 14])
```

#### Operator based question

##### The rule for `and` is:

- If the first operand is truthy, return the second operand.
- If the first operand is falsy, return the first operand

##### The rule for `or` is:

- If the first operand is falsy, return the second operand.
- If the first operand is truthy, return the first operand.

```python
result = [1, 2] and "Hello"
print(result)
```

```python
result = 0 and 1 / 0
print(result)
```

```python
result = [] or "Hello"
print(result)
```

```python
result = 0 or (1, "hi", "dhyan")
print(result)
```

#### Slicing in Strings

- Example 1

```python
x="be active in class"

#active
p1 = x[3:9:1]
print(p1)
n1 = x[-15:-9:1]
print(n1)

#in
p2 = x[10:12:1]
print(p2)
n2 = x[-8: -6: 1]
print(n2)

#lass
p3 = x[14: 18: 1]
print(p3)
n3 = x[-4 : 18: 1]
print(n3)

#be
p4 = x[0: 2: 1]
print(p4)
n4 = x[-18: -16: 1]
print(n4)

#atv
p5 = x[3 : 8 : 2]
print(p5)
n5 = x[-15:-9: 2]
print(n5)
```

#### Interview Questions regarding String datatype

##### what is the difference between find and index method
- index: it will return the position/index of a specified value, also return value error if the specified value is not present.
- find: it will return the position/index of a specified value, will return -1 if the specified value is not present.

##### what is the difference between indexing and slicing
- indexing: will return a single element at a specific position
- slicing: will return a portion/sequence of elements from a range of positions.

##### what is the meaning of mutable data type with example
- mutable data type can be modified/changed after creation, example: list, we can add, remove, or change elements in a list after it is created.

##### why we are using triple quotes in string data type
- it is used to create multi-line strings and to define docstrings in python.

##### what is the difference between split( ) method and  rsplit( )
- `split()` will split the string from the left side, 
- `rsplit()` will split the string from the right side, 
- both behave same without maxsplit parameter but differ when maxsplit is specified.

##### what is the use of dir ( ) function
- it will return a list of all valid attributes and methods of the specified object.

##### how to create an empty string using object
- we can create an empty string by calling str() with no arguments, like `s = str()`, or by assigning empty quotes like `s = ""`.

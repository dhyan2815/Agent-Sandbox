#### Find all duplicate elements in a list.

````python
def find_duplicates(arr):
    seen = []
    duplicates = []
    for item in arr:
        if item in seen and item not in duplicates:
            duplicates.append(item)
        else:
            seen.append(item)
    return duplicates

list = [1, 2, 3, 4, 1, 2, 2, 5, 5, 4]
dupli_list = find_duplicates(list)

# Sort the list in ascending order
sorted(dupli_list)
````

#### Count the frequency of each Letter

- Using Library

````python
from collections import Counter

word = "The word taken is Banana"

count = Counter(word.lower())
print(count)
````

- Using without Library

````python
word = 'Banana'
letter_count = {}

for letter in word:
    if letter in letter_count:
        letter_count[letter] += 1
    else:
        letter_count[letter] = 1
print(letter_count)
````

#### Given a list of numbers, return a list of the squares.

````python
def squares(num):
    return [x**2 for x in num]
squares([1, 2, 3, 5, 6])
````

#### Given a list of numbers, return a list of the squares of only the even numbers.

````python
def sq_num(no):
    return [x**2 for x in no if x % 2 == 0]

sq_num([1, 2, 3, 4, 5, 6, 8, 10, 12, 14])
````

## Flattened Nested List

- Write a function to flatten a nested list into 1d list.

````python
def flatten_list(lst):
  flat_list = []

  for item in lst:

    if isinstance(item, list):
      flat_list.extend(item)

    else:
      flat_list.append(item)

  return flat_list

print(flatten_list([1, [2, 3, 4], [5, 6], 7, [9, 10]]))
````

## List Intersection
- Given two list, return a list of elements that exist in both

````python
def resultant_list(l1, l2):
  return list(set(l1) & set(l2))

l1 = [1, 2, 3]
l2 = [2, 5, 3]

print(resultant_list(l1, l2))
````

## Second Largest
- Find the second largest number in a list without using sort().


````python
def get_sec_largest(num):
  unique_nums = set(num)

  if len(unique_nums) < 2:
    return None

  unique_nums.remove(max(unique_nums))
  return max(unique_nums)

print(get_sec_largest([1, 2, 3, 1, 1, 2, 2, 79, 79]))

# we used a logic in which we first converted the list into set and then used set built-in function to remove the
# largest number using remove and then print the latest max number in list,
# which is practically the second largest number is what we have printed.
````

### 3. Find Duplicate Characters in a String

s = "programming"
- Output: `r, g, m`

````python
def duplicates_chars():

  s = "programming"

  # count occurence of each character
  char_count = {}
  for char in s:
    char_count[char] = char_count.get(char, 0) + 1

  # find characters that appears more that once
  duplicates = []
  for char, count in char_count.items():
    if count > 1:
      duplicates.append(char)
  return ", ".join(duplicates)

duplicates_chars()
````

#### to find **Duplicate Characters in String** in-built functions where used

1. `get()` : *dict* datatype - fetches specified value of specified key
2. `items()`: *dict* datatype - fetches all key value pair
3. `append()`: *list* datatype - adds char/element to the list

### 6. Reverse Words in a Sentence
s = "I love Python"
```
# Output: Python love I
```

````python
s  = "I love Python"

def reverse_words(s):

  words = s.split() # split string into list seperating each word as a single element
  reverse_list = words[::-1] # used reverse indexing to reverse the order of the word
  result = " ".join(reverse_list) # used whitespace to contain a space between each word store this into a string

  print(result)

reverse_words(s)
````

### If-Else Statement Questions

- WAP to check number is even number.

```python
num = eval(input("Enter number: "))

if (num%2==0):
  print(f"{num} is even")
else:
  print(f"{num} is odd")
```

- WAP to take one character from user and if ASCII value of the character is odd then make it as even ASCII and convert to character else if ASCII value is already even then make it as odd and covert to character.

```python
ch = input("Enter character: ")
asc = ord(ch)

if (asc % 2 != 0):
  asc = chr(asc + 1)
  print(asc)
else:
  asc = chr(asc + 1)
  print(asc)
```

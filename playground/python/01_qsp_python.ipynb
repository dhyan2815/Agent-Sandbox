6. wap to take marks of 5 sub,calculate the average if the average is b/w 90-100 print Distinction

note:-->max marks is 100

```python
m1 = eval(input("Enter eng marks: "))
m2 = eval(input("Enter sci marks: "))
m3 = eval(input("Enter pt marks: "))
m4 = eval(input("Enter hindi marks: "))
m5 = eval(input("Enter guj marks: "))
m6 = eval(input("Enter social marks: "))

def avg(n1, n2, n3, n4, n5):
  total_sum = n1 + n2 + n3 + n4 + n5
  return total_sum / 5

if (90 <= avg(m1, m2, m3, m4, m5) <= 100):
  print("Distinction")
elif (75 <= avg(m1, m2, m3, m4, m5) <= 89):
  print("First Class")
elif (60 <= avg(m1, m2, m3, m4, m5) <= 74):
  print("Second Class")
elif (50 <= avg(m1, m2, m3, m4, m5) <= 59):
  print("third Class")
elif (avg(m1, m2, m3, m4, m5) < 50):
  print("fail")
```

10. Tax Slab Calculator

Write a program that takes annual income as input and calculates tax:

Up to ₹2,50,000 → No tax

₹2,50,001–₹5,00,000 → 5%

₹5,00,001–₹10,00,000 → 20%

Above ₹10,00,000 → 30%

```python
an_in = eval(input("Enter annual income: "))

if (an_in <= 250000):
  tax_per = 0
  tax_amt = an_in + tax_per
  total_slab = an_in + tax_amt
  print(f"Total Slab: {total_slab}, Tax (%): {tax_per}")
elif (250001 <= an_in <= 500000):
  tax_per = 0.05
  tax_amt = an_in * tax_per
  total_slab = an_in + tax_amt
  print(f"Total Slab: {total_slab}, Tax (%): {tax_per}")
elif (500001 <= an_in <= 1000000):
  tax_per = 0.20
  tax_amt = an_in * tax_per
  total_slab = an_in + tax_amt
  print(f"Total Slab: {total_slab}, Tax (%): {tax_per}")
elif (an_in < 1000000):
  tax_per = 0.30
  tax_amt = an_in * tax_per
  total_slab = an_in + tax_amt
  print(f"Total Slab: {total_slab}, Tax (%): {tax_per}")
else:
  print("Invalid annual income")
```

11. E-commerce Delivery Charges

 A shopping site charges delivery fees:

Order < ₹500 → ₹50 charge

₹500–₹1000 → ₹30 charge

₹1000–₹2000 → Free delivery

Above ₹2000 → Free delivery + 5% cashback

```python
order = eval(input("Enter order amount: "))

print("\nNote: TOV = Total Order Value\n")

if (order < 500):
  charge = 50
  total_amt = order + charge
  print(f"TOV: {total_amt}")
elif (500 < order < 1000):
  charge = 30
  total_amt = order + charge
  print(f"TOV: {total_amt}")
elif (1000 < order < 2000):
  charge = 0
  total_amt = order + charge
  print(f"TOV: {total_amt}")
elif (order > 2000):
  charge = 0
  cashback = (order * 5) / 100
  total_amt = (order + charge) - cashback
  print(f"TOV: {total_amt}")
```

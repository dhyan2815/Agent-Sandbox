### Simple If Statement Questions

- WAP to check number is odd number.

```python
num = eval(input("Enter a number: "))

if (num % 2 != 0):
  print(f"{num} is odd")
```

- WAP to check number is even number.

```python
num = eval(input("Enter a number: "))

if (num % 2 == 0):
  print(f"{num} is even")
```

- WAP to check character is vowel.

```python
ch = input("Enter character: ")

if ch in "AEIOUaeiou":
  print("Character is Vowel")
```

- WAP to check character is not vowel (consonant).

```python
ch = input("Enter character")

if ch not in "AEIOUaeiou":
  print("Char is consonant")
```

- WAP to take one dictionary from user and check if the last value of dictionary is odd then print the square of that value, else add one key-value pair in dictionary.

```python
d = eval(input("Enter dictionary: "))

removed_el = d.popitem()

if (removed_el[-1] % 2 != 0):
  print(removed_el[-1]**2)
else:
  d.update({12: 23})
  print(d)
```

19. wap to check whether a character is in the alphabet or not,if it is alphabet,store the value inside  a dict(key as a character and value as a ascii value)

```python
ch = input("Enter character: ")

if (ch.isalpha()):
  d = {}
  d.setdefault(ch, ord(ch))
  print(d)
else:
  print("Char is not alphabet")
```

20. wap to check whether a character is in uppercase or not,if uppercase,convert to lowercase and store the value inside the dictionary (character as key and ascii as value) take user input

```python
ch = input("Enter character: ")

if (ch.isupper()):
  lowercased_ch = ch.swapcase()
  d = {}
  d.setdefault(lowercased_ch, ord(lowercased_ch))
  print(d)
else:
  print("Char is not uppercase")
```
